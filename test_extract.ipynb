```python
import re
import pandas as pd
from pathlib import Path

file_path = Path('gempa.txt')

text = file_path.read_text()
```

```python
# Pola untuk memisahkan per tanggal
date_pattern = r"(?:Senin|Selasa|Rabu|Kamis|Jumat|Sabtu|Minggu), \d{2} \w+ \d{4}"
entries = re.split(date_pattern, text)[1:]
dates = re.findall(date_pattern, text)

# Pola untuk setiap baris gempa
gempa_pattern = r"(\d+) kali(?: gempa)? ([\w\s]+) dengan amplitudo ([\d\.]+(?:-[\d\.]+)?) mm.*?lama gempa ([\d\.]+(?:-[\d\.]+)?) detik"

# Ekstraksi data ke list
data = []
for date, entry in zip(dates, entries):
    matches = re.findall(gempa_pattern, entry)
    for jumlah, nama, amplitudo, durasi in matches:
        amplitudo_min = None
        amplitudo_max = amplitudo
        durasi_min = None
        durasi_max = durasi
        if int(jumlah) > 1:
            amps = amplitudo.split('-')
            print(date.split(', ')[1], amps)
            if len(amps) == 2:
                [amplitudo_min, amplitudo_max] = amps
            else:
                amplitudo_min = None
                amplitudo_max = amps[0]
            durations = durasi.split('-')
            if len(durations) == 2:
                [durasi_min, durasi_max] = durations
            else:
                durasi_min = None
                durasi_max = durations[0]
        data.append({
            "Tanggal": date.split(', ')[1],
            "Nama Gempa": nama.strip(),
            "Jumlah": int(jumlah),
            "Amplitudo Min (mm)": float(amplitudo_min) if amplitudo_min is not None else None,
            "Amplitudo Max (mm)": float(amplitudo_max),
            "Durasi Min (detik)": float(durasi_min) if durasi_min is not None else None,
            "Durasi Max (detik)": float(durasi_max)
        })

# Buat DataFrame
df = pd.DataFrame(data)
```

```python
import pandas as pd
import re

# Data gempa yang akan diolah
data_raw = """
Selasa, 06 Mei 2025, terekam 36 kali gempa Hembusan dengan amplitudo 2.2-10.5 mm, dan lama gempa 24-235 detik.
12 kali Harmonik dengan amplitudo 2.2-7.4 mm, dan lama gempa 31-257 detik.
2 kali gempa Low Frequency dengan amplitudo 4.4-7.4 mm, dan lama gempa 28-37 detik.
7 kali gempa Vulkanik Dalam dengan amplitudo 2.9-14.8 mm, S-P 1.3-2.7 detik dan lama gempa 10-21 detik.
2 kali gempa Tektonik Lokal dengan amplitudo 5.9-22.2 mm, S-P 4.7-8 detik dan lama gempa 18-29 detik.
3 kali gempa Tektonik Jauh dengan amplitudo 4.4-7.4 mm, S-P 14.1 detik dan lama gempa 46-145 detik.
Rabu, 07 Mei 2025, terekam 22 kali gempa Hembusan dengan amplitudo 2.2-14.8 mm, dan lama gempa 24-221 detik.
16 kali Harmonik dengan amplitudo 2.9-7.4 mm, dan lama gempa 74-280 detik.
1 kali gempa Low Frequency dengan amplitudo 4.4 mm, dan lama gempa 29 detik.
4 kali gempa Vulkanik Dalam dengan amplitudo 4.4-7.4 mm, S-P 1-1.7 detik dan lama gempa 10-16 detik.
3 kali gempa Tektonik Lokal dengan amplitudo 5.9-29.6 mm, S-P 4.2-5.7 detik dan lama gempa 20-26 detik.
7 kali gempa Tektonik Jauh dengan amplitudo 2.2-11 mm, S-P 11.7-37 detik dan lama gempa 43-193 detik.
Kamis, 08 Mei 2025, terekam 11 kali gempa Hembusan dengan amplitudo 2.2-5.9 mm, dan lama gempa 30-147 detik.
19 kali Harmonik dengan amplitudo 2.9-7.4 mm, dan lama gempa 40-273 detik.
2 kali gempa Low Frequency dengan amplitudo 2.9-4.4 mm, dan lama gempa 20-23 detik.
8 kali gempa Vulkanik Dalam dengan amplitudo 2.9-7.4 mm, S-P 1.1-1.9 detik dan lama gempa 9-16 detik.
2 kali gempa Tektonik Lokal dengan amplitudo 4.4-14.8 mm, S-P 4.7-8.1 detik dan lama gempa 20-46 detik.
4 kali gempa Tektonik Jauh dengan amplitudo 2.9-29.6 mm, S-P 13.4-21 detik dan lama gempa 39-79 detik.
"""

# Fungsi untuk parsing data gempa
def parse_earthquake_data(data_text):
    lines = data_text.strip().split("\n")
    
    results = []
    current_date = None
    
    for line in lines:
        # Cek apakah baris ini berisi tanggal
        date_match = re.search(r"(Senin|Selasa|Rabu|Kamis|Jumat|Sabtu|Minggu), (\d{2} \w+ \d{4})", line)
        if date_match:
            current_date = date_match.group(2)
        
        # Jika tidak ada tanggal yang ditetapkan, lewati baris ini
        if not current_date:
            continue
        
        # Ekstrak jumlah dan jenis gempa
        count_match = re.search(r"(\d+) kali gempa (\w+(?:\s+\w+)*?) dengan", line)
        if not count_match:
            count_match = re.search(r"terekam (\d+) kali gempa (\w+(?:\s+\w+)*?) dengan", line)
        
        if count_match:
            count = int(count_match.group(1))
            quake_type = count_match.group(2)
            
            # Ekstrak amplitudo
            amp_match = re.search(r"amplitudo (\d+\.\d+)(?:-(\d+\.\d+))? mm", line)
            if amp_match:
                if amp_match.group(2):  # Jika ada rentang nilai
                    amp_min = float(amp_match.group(1))
                    amp_max = float(amp_match.group(2))
                else:  # Jika hanya satu nilai
                    amp_min = None
                    amp_max = float(amp_match.group(1))
            else:
                amp_min = None
                amp_max = None
            
            # Ekstrak durasi gempa
            duration_match = re.search(r"lama gempa (\d+)(?:-(\d+))? detik", line)
            if duration_match:
                duration_min = int(duration_match.group(1))
                duration_max = int(duration_match.group(2)) if duration_match.group(2) else duration_min
            else:
                duration_min = None
                duration_max = None
            
            # Tambahkan ke hasil
            results.append({
                'Tanggal': current_date,
                'Jenis Gempa': quake_type,
                'Jumlah': count,
                'Amplitudo Min (mm)': amp_min,
                'Amplitudo Max (mm)': amp_max,
                'Durasi Min (detik)': duration_min,
                'Durasi Max (detik)': duration_max
            })
    
    return results

# Parse data
earthquake_data = parse_earthquake_data(data_raw)

# Buat DataFrame
df = pd.DataFrame(earthquake_data)

# Format tabel sesuai permintaan
table_df = df.copy()
table_df['Amplitudo'] = table_df.apply(
    lambda row: f"{row['Amplitudo Min (mm)']}" if row['Amplitudo Min (mm)'] == row['Amplitudo Max (mm)'] 
    else f"{row['Amplitudo Min (mm)']} - {row['Amplitudo Max (mm)']}", axis=1
)
table_df['Durasi'] = table_df.apply(
    lambda row: f"{row['Durasi Min (detik)']}" if row['Durasi Min (detik)'] == row['Durasi Max (detik)'] 
    else f"{row['Durasi Min (detik)']} - {row['Durasi Max (detik)']}", axis=1
)

# Pilih kolom yang diperlukan untuk output
output_df = table_df[['Tanggal', 'Jenis Gempa', 'Jumlah', 'Amplitudo Min (mm)', 'Amplitudo Max (mm)', 'Durasi']]

# Ubah nama kolom sesuai permintaan
output_df = output_df.rename(columns={
    'Tanggal': 'Tanggal',
    'Jenis Gempa': 'Nama Gempa',
    'Jumlah': 'Jumlah Gempa',
    'Amplitudo Min (mm)': 'Amplitudo Minimum',
    'Amplitudo Max (mm)': 'Amplitudo Maksimum',
    'Durasi': 'Durasi'
})

# Tampilkan tabel
print("\nTabel Data Gempa:")
print(output_df.to_string(index=False))

# Simpan ke CSV (opsional)
output_df.to_csv('data_gempa.csv', index=False)
print("\nData telah disimpan ke file 'data_gempa.csv'")
```
