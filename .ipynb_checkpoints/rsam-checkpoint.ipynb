```python
from magma_rsam import PlotRsam

plot_rsam = PlotRsam(
    start_date="2026-01-01",
    end_date="2026-01-31",
    station="DKB",
    channel="EHZ",
    network="VG",
    location="00",
    resample="10min",
    rsam_dir=r"G:\DUKONO\output\rsam"
).with_filter(
    freq_min=8.0,
    freq_max=16.0
)

print(plot_rsam.nslc)
print(plot_rsam.rsam_dir)
print(plot_rsam.filtered_dir)
print(plot_rsam.csv_files)
print("Jumlah csv_files:", len(plot_rsam.csv_files))
```

```python
from magma_database import RsamCSV

rows = list(RsamCSV.select().where(RsamCSV.nslc == "VG.DKB.00.EHZ"))

print("Jumlah record untuk VG.DKB.00.EHZ:", len(rows))

for r in rows[:50]:
    print(
        "date=", r.date,
        "resample=", r.resample,
        "freq_min=", r.freq_min,
        "freq_max=", r.freq_max,
        "file=", r.file_location
    )
```
