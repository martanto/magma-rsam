```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Generate hourly timestamps from June 6 to June 16, 2020
date_range = pd.date_range(start='2020-10-01', end='2020-12-31', freq='24h')
n = len(date_range)
```

```python
# Generate base consensus values
np.random.seed(0)
consensus = np.random.normal(0.4, 0.05, n)
```

```python
consensus
```

```python

# Simulate a rise to eruption starting around June 14 23:00 (index 215)
start_rise_idx = 215
end_rise_idx = 240  # About 25 hours later
consensus[start_rise_idx:end_rise_idx] = np.linspace(0.6, 1.0, end_rise_idx - start_rise_idx)
consensus[end_rise_idx:] = 0.98  # Sustained high consensus

# Threshold value
threshold = 0.7

# Define eruption window and eruption time
start_ep = pd.Timestamp("2020-11-30 00:00")
end_ep = pd.Timestamp("2020-12-04 00:00")
eruption_time = end_ep

# Plotting
fig, ax = plt.subplots(figsize=(10, 2.5))

# Consensus line
ax.plot(date_range, consensus, color='black', linewidth=1, label='Consensus')

# Threshold line
ax.axhline(threshold, color='orange', linestyle='--', linewidth=2, label='Threshold')

# Shaded eruptive episode
ax.axvspan(start_ep, end_ep, color='red', alpha=0.2, label='Eruptive episode')

# Vertical line at eruption
ax.axvline(eruption_time, color='red', linewidth=2)

# Duration annotation
mid_point = start_ep + (end_ep - start_ep) / 2
ax.annotate("~25 hr", xy=(mid_point, 1.02), xytext=(mid_point, 1.07),
            ha='center', fontsize=10, color='blue',
            arrowprops=dict(arrowstyle='<->', color='black'))

# Labels and title
ax.set_ylabel("Consensus")
ax.set_xlabel("Time")
ax.set_ylim(0, 1.1)
ax.set_title("Eruptive episode 1")

# Legend
ax.legend(loc='upper left')

# Format and show
plt.tight_layout()
plt.show()
```
