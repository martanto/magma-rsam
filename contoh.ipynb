```python
from magma_rsam import RSAM, PlotRsam
import matplotlib.pyplot as plt
from datetime import datetime
import pandas as pd
```

```python
start_date='2023-06-19'
end_date='2023-06-20'
station="AWU1"
channel="EHZ"
```

**) Depends on time. For Dieng, SAC was used from 2013-08-12 to 2021-09-15 and 2023-10-16 to 2024-08-15. 
Seisan was used from 2021-03-26 to 2023-10-16. 

```python
rsam = RSAM(
    #liat tabel di atas kolom directory_structure
    directory_structure='awu', 
    
    # tempat data seismic disimpan
    # jangan lupa double backslash \\ buat misahin foldernya
    seismic_dir="D:\\Temp\\Sample_awu",
    station=station,
    channel=channel,
    start_date=start_date,
    end_date=end_date,
)
```

Dikomen aja pake `#` kalau gak mau pake filter. Atau di skip aja 

```python
freq_min = 1.0
freq_max = 5.0
rsam.apply_filter(freq_min=freq_min, freq_max=freq_max)
```

Kalkulasi RSAM

```python
rsam.run()
```

Plotting

```python
plot_rsam = PlotRsam(
    station=station,
    channel=channel,
    start_date=start_date,
    end_date=end_date,
)
```

Kalau mau plot pake filter. Pastikan RSAM nya udah di run pake filter juga. Boleh diskip

```python
plot_rsam.with_filter(freq_min=freq_min, freq_max=freq_max)
```

```python
plot_rsam.filtered_dir
```
