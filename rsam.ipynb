```python
from magma_rsam import RSAM, PlotRsam
from magma_converter.database import DatabaseConverter
```

```python
rsam = RSAM(
    seismic_dir="G:\\DUKONO\\SDS",
    station='DKB',
    channel='EHZ',
    network = 'VG',
    location = '00',
    start_date = '2026-01-01',
    end_date = '2026-01-31',
    directory_structure = 'sds',
    update_db=True
)
```

```python
rsam.apply_filter(freq_min=8.0, freq_max=16.0)
```

```python
DatabaseConverter.check_tables()
rsam.run()
```

```python
plot_rsam = PlotRsam(
    station = 'DKB',
    channel = 'EHZ',
    network = 'VG',
    location = '00',
    start_date = '2026-01-01',
    end_date = '2026-01-31',
    resample = '10min'
)
```

```python
plot_rsam.with_filter(freq_min=8.0, freq_max=16.0)
```

```python
plot_rsam.set_y_lim(
        y_min=0, 
        y_max=750
    ).run(
        datetime_interval=14,
    )
```
